```python
# Modelo de classes

class Pessoa:
    def __init__(self, nome, idade, nacionalidade):
        self.nome = nome
        self.idade = idade
        self.nacionalidade = nacionalidade

    def cumprimentar(self):
        return f"Olá, meu nome é {self.nome} e tenho {self.idade} anos."
    
    def fazAniversario(self):
        self.idade += 1
        return f"feliz aniversário, {self.nome}! Agora você tem {self.idade} anos"
    
    def pais(self):
                
        if self.nacionalidade == 1:
            return f"Nacionalidade Brasileira {self.nacionalidade}"
        else:
            return f"Nacionalidade Estrangeira {self.nacionalidade}"
        
    # criando um obejto da classe Pessoa

pessoa1 = Pessoa("Maria", 25, 1)
pessoa2 = Pessoa("José", 53, 2)


print(pessoa1.cumprimentar())
print(pessoa1.fazAniversario())
print(pessoa1.pais())

print(pessoa2.cumprimentar())
print(pessoa2.fazAniversario())
print(pessoa2.pais())
```

```python
# exercício de classes:

class Produto:
    def __init__(self, nome, preco, qtde_estoque):
        self.nome = nome
        self.preco = preco
        self.qtde = qtde_estoque

    def exibirDados(self):
        return self.nome, self.preco, self.qtde
    

    def vender(self, qtde):
        if self.qtde >= qtde:
            self.qtde = self.qtde - qtde
            print(f"O estoque agora é: {self.qtde}")
        else:
            print("Produto sem estoque")
    def repor(self, qtde):
        self.qtde = self.qtde + qtde
        print(f"O estoque agora é: {self.qtde}")

    

nome_produto = str(input("DIgite o nome do produto: "))
est_produto = int(input("Digite o estoque do produto: "))
preco_produto = float(input("Digite o preço do produto"))

produto1 = Produto(nome_produto, preco_produto, est_produto)
print(produto1.exibirDados())
produto1.vender(3)
print(produto1.exibirDados())
produto1.repor(4)
print(produto1.exibirDados())
print("________________________________________________")



```

```python
# resolução do professor
class Veiculo():
    def __init__(self, modelo, combustivel):
        self.modelo = modelo
        self._combustivel = combustivel

    def exibir_info(self):
        print(f"Modelo: {self.modelo} - Combustível {self._combustivel}")
    
    def abastecer(self, litros):
        if litros > 0:
            self._combustivel += litros
        else:
            print("Quantidade inválida!")
    def get_combustivel(self):
        return self._combustivel

class Carro(Veiculo):
    def abastecer(self, litros):
        super().abastecer(litros)
        print(f"Carro {self.modelo} foi abastecido com {litros}")

class Moto(Veiculo):
    def abastecer(self, litros):
        super().abastecer(litros)
        print(f"Carro {self.modelo} foi abastecido com {litros}")

moto1 = Moto("Biz", 0)
carro1 = Carro("Pegeout", 0)

moto1.exibir_info()
carro1.exibir_info()

moto1.abastecer(3)
carro1.abastecer(40)


moto1.exibir_info()
carro1.exibir_info()
```

```python
# ABSTRAÇÃO
from abc import ABC, abstractmethod

class Forma(ABC):
    @abstractmethod
    def calcular_area(self):
        pass

class Retangulo(Forma):
    def __init__(self, altura, largura):
        self.altura = altura
        self.largura = largura
        
    def calcular_area(self):
        return self.altura * self.largura

class Ciruculo(Forma):
    def __init__(self, raio):
        self.raio = raio
        
    def calcular_area(self):
        return self.raio * 2 * 3.14
    
class Losango(Forma):
    def __init__(self, diagonal1, diagonal2):
        self.diagonal1 = diagonal1
        self.diagonal2 = diagonal2

    def calcular_area(self):
        return (self.diagonal1 * self.diagonal2) /2


formas = [Retangulo(4,3), Ciruculo(4), Losango(4,5)]

for forma in formas:
    print( forma.calcular_area())
```

```python
# Exercicio polimorfismo
'''
Requisitos:
1. Crie uma classe base abstrata chamada Pagamento, que terá:
o Um atributo valor (float), representando o valor da compra.
o Um método processar_pagamento(), que será abstrato e deverá ser
sobrescrito pelas subclasses.

2. Crie as subclasses:
o CartaoCredito:
▪ Atributos adicionais: numero_cartao (str) e cvv (str).
▪ O método processar_pagamento() deve:
▪ Verificar o número do cartão.
▪ Autenticar com o CVV.
▪ Exibir uma mensagem de aprovação.

o BoletoBancario:
▪ Atributo adicional: cpf_cliente (str).
▪ O método processar_pagamento() deve:
▪ Gerar o boleto com vencimento de 3 dias.
▪ Mostrar o CPF do cliente.


o Pix:
▪ Atributo adicional: chave_pix (str).
▪ O método processar_pagamento() deve:
▪ Exibir a chave Pix usada.
▪ Confirmar que o pagamento foi instantâneo.


3. Crie uma lista contendo objetos das diferentes formas de
pagamento.
o Cada objeto deve ser instanciado com valores fictícios e
realistas.
o Em seguida, use um laço for para processar todos os
pagamentos, demonstrando o polimorfismo.

'''
from abc import ABC, abstractmethod
from random import randint

class Pagamento(ABC):
    def __init__(self, valor):
        self.valor = valor
    @abstractmethod
    def processar_pagamento(self, valor):
        pass

class Cartao_Credito(Pagamento):
    def __init__(self, numero_cartao, ccv, valor):
        super().__init__(valor)
        self.valor = valor
        self.numero_cartao = numero_cartao
        self.ccv = ccv

    def processar_pagamento(self):
        if len(self.numero_cartao) == 16 and len(self.ccv) == 3:
            print(f"Cartão de crédito validado: {self.numero_cartao}, Pagamento aprovado")
        else:
            print("Número de cartão inválido!")
    

class BoletoBancario(Pagamento):
    def __init__(self, cpf_cliente, valor):
        super().__init__(valor)
        self.valor = valor
        self.cpf_cliente = cpf_cliente
    
    def processar_pagamento(self):
        if len(self.cpf_cliente) == 11:
            print(f"Boleto gerado, vence em três dias, para o cpf {self.cpf_cliente}")

class Pix(Pagamento):
    def __init__(self, chave_pix, valor):
        super().__init__(valor)
        self.valor = valor
        self.chave_pix = chave_pix

    def processar_pagamento(self):
        print("Pagamento estantâneo efetuado")


formas_pagamento = [Cartao_Credito("1234567891234896", "123",1), BoletoBancario("65496213568",1), Pix("56kjhgd",1)]


for formas in formas_pagamento:
    formas.processar_pagamento()




        


```
