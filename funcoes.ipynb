```python
def soma (a,b):
    return a + b

def subtracao (a,b):
    subtrair = a - b
    return subtrair

def divisao (a,b):
    dividir = a / b
    return dividir

def multiplicacao (a,b):
    multiplicar = a * b
    return multiplicar



opcao = int(input("Digite uma opção: "))
print(opcao)

while opcao != 0:
    c = int(input("Digite um número"))
    d = int(input("Digite outro número"))
    if opcao == 1:
        print(soma(c,d))

    if opcao == 2:
        print(subtracao(c,d))

    if opcao == 3:
        print(divisao(c,d))

    if opcao == 4:
        print(multiplicacao(c,d))




```
