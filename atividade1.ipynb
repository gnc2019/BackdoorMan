#### 4 - Escreva um algoritimo que leia a temperatura em graus Celsius e o estado do tempo (ensolarado, nublado ou chuvoso) e classificar o clima com base nas seguintes condições:

#### Se a temperatura < 0 E estado = __nublado__ -> Clima extremo: frio e nublado;
#### Se a temperatura >= 0 E temperatura < 15 E estado = chuvoso ou nublado -> Muito Frio e úmido
#### Se a temperatura >= 15 E temperatura < 20 E estado = ensolarado -> Frio com nuvens

#### Se a temperatura >= 20 E temperatura < 25 E estado # ensolarado -> lima agradável e ensolarado
#### Se a temperatura >= 25 E temperatura < 30 E estado # chuvoso -> Quente e seco
#### Se temperatura >= 30 E estado = ensolarado -> Muito quente e ensolarado - cuidado com o calor!
#### Se nehuma condição for atendia -> Clima indefinido


```python
# Se a temperatura < 0 E estado = nublado -> Clima extremo: frio e nublado;
temperatura = int(input("Digite a temperatura atual: "))
estado = input("Digite a estado atual: ")

    
print(temperatura, estado)

if temperatura < 0 and estado == "nublado":
    print("frio e nublado")

elif temperatura >= 0 and temperatura < 15 or estado == "chuvoso" or estado == "nublado":
    print("Muito Frio e úmido")

elif temperatura >=15 and temperatura < 20 and estado == "nublado":
    print("Frio com nuvens")

elif temperatura >=20 and temperatura < 25 and estado == "ensolarado":
    print("Clima agradável e ensolarado")

elif temperatura >=25 and temperatura < 30 != estado == "chuvoso":
    print("Quente e seco")

elif temperatura >=30 and estado == "ensolarado":
    print("Muito quente e ensolarado - cuidado com o calor!")

else:
    print("Clima indefinido")
```

```python
# O sript deve pedir o valor de cada produto comprado
# a cada produto o valor digitado deve ser adicionado ao total da compra.
# qaundo o clinte não tiver mais produtos ele deve digitar 0 para finalizar a compra
# deve mostrar o total da compra
# quantidade de produtos comprados
# 

qtde = 0
soma_total = 0
produto = float(input("Digite o valor do produto"))

while produto != 0:
    qtde = qtde + 1
    soma_total = soma_total + produto
    produto = float(input("Digite o valor do produto"))

print("Quantidade: {}, Soma total da compra: R$ {:.2f}".format(qtde, soma_total))
type(soma_total)      
```

```python
# Método pop()
# pop(): Remove e retorna o item da lista em uma posição específica (ou do final, por padrão).
lista = [1, 2, 3, 4]
item = lista.pop(1)  # Remove o item na posição 1 (valor 2)
print(lista)  # Saída: [1, 3, 4]
print(item)  # Saída: 2
```
